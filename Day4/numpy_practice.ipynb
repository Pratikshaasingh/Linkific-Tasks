```python
import numpy as np

# 1D Array 
arr_1d = np.array([85, 92, 78, 64, 90])

# 2D Array 
arr_2d = np.array([
    [85, 92, 78, 88],
    [64, 70, 68, 75],
    [90, 95, 91, 89]
])

print("1D Array Shape:", arr_1d.shape)  
print("2D Array Shape:", arr_2d.shape)  
```

```python
# 1. Slice 1D array 
slice_1d = arr_1d[1:4]  

# 2. Slice 2D array 

sub_matrix = arr_2d[0:2, 1:3]

# 3. Boolean Indexing 
high_scores = arr_1d[arr_1d >= 85]

print("Sub-matrix:\n", sub_matrix)
print("Scores >= 85:", high_scores)
```

```python
# Overall metrics
print("Total Sum:", np.sum(arr_2d))
print("Overall Mean:", np.mean(arr_2d))
print("Max Value:", np.max(arr_2d))
print("Min Value:", np.min(arr_2d))

print("Mean per column (subject):", np.mean(arr_2d, axis=0))
print("Mean per row (student):", np.mean(arr_2d, axis=1))
```

```python
import numpy as np

marks = np.array([
    [85, 78, 92, 88],
    [64, 70, 68, 75],
    [90, 95, 91, 89],
    [55, 60, 58, 62],
    [78, 82, 80, 85]
])

subjects = ["Math", "Science", "English", "History"]

# Calculate metrics
student_averages = np.mean(marks, axis=1)
subject_averages = np.mean(marks, axis=0)

print(" Student Averages ")
for i, avg in enumerate(student_averages):
    print(f"Student {i+1}: {avg:.2f}")

print("\n Subject Averages ")
for sub, avg in zip(subjects, subject_averages):
    print(f"{sub}: {avg:.2f}")
```
