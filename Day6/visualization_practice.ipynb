```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Set visual style
sns.set_theme(style="whitegrid")

# Load the Day 5 Iris dataset
df = pd.read_csv("../Day5/iris.csv")
```

1. Bar Chart (Average Petal Length per Species)

```python
plt.figure(figsize=(8, 5))
avg_petal = df.groupby('species')['petal_length'].mean().reset_index()

sns.barplot(data=avg_petal, x='species', y='petal_length', palette='viridis')
plt.title('Average Petal Length by Iris Species', fontsize=14, fontweight='bold')
plt.xlabel('Species', fontsize=12)
plt.ylabel('Mean Petal Length (cm)', fontsize=12)
plt.savefig('bar_chart.png', dpi=300, bbox_inches='tight')
plt.show()
```

2. Line Chart (Trend of Sepal Length Across Sample Index)

```python
plt.figure(figsize=(10, 5))
plt.plot(df.index, df['sepal_length'], color='darkblue', linewidth=1.5, label='Sepal Length')

plt.title('Sepal Length Trend Across Dataset Index', fontsize=14, fontweight='bold')
plt.xlabel('Sample Index', fontsize=12)
plt.ylabel('Sepal Length (cm)', fontsize=12)
plt.legend(loc='upper left')
plt.savefig('line_chart.png', dpi=300, bbox_inches='tight')
plt.show()
```

3. Histogram (Distribution of Sepal Width)

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['sepal_width'], kde=True, bins=15, color='teal')

plt.title('Distribution of Sepal Width', fontsize=14, fontweight='bold')
plt.xlabel('Sepal Width (cm)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.savefig('histogram.png', dpi=300, bbox_inches='tight')
plt.show()
```

4. Pie Chart (Proportion of Species Samples)

```python
plt.figure(figsize=(6, 6))
species_counts = df['species'].value_counts()

plt.pie(species_counts, labels=species_counts.index, autopct='%1.1f%%', 
        colors=['#66b3ff','#99ff99','#ffcc99'], startangle=140, explode=(0.05, 0.05, 0.05))
plt.title('Species Sample Proportion in Iris Dataset', fontsize=14, fontweight='bold')
plt.savefig('pie_chart.png', dpi=300, bbox_inches='tight')
plt.show()
```

1. Scatter Plot with Species 

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='sepal_length', y='petal_length', hue='species', style='species', s=70, palette='Set2')

plt.title('Sepal Length vs. Petal Length by Species', fontsize=14, fontweight='bold')
plt.xlabel('Sepal Length (cm)', fontsize=12)
plt.ylabel('Petal Length (cm)', fontsize=12)
plt.legend(title='Species', loc='upper left')
plt.savefig('scatterplot.png', dpi=300, bbox_inches='tight')
plt.show()
```

2. Box Plot (Outlier & Distribution Analysis)

```python
plt.figure(figsize=(9, 5))
sns.boxplot(data=df, x='species', y='sepal_width', palette='Pastel1')

plt.title('Sepal Width Distribution & Outlier Analysis per Species', fontsize=14, fontweight='bold')
plt.xlabel('Species', fontsize=12)
plt.ylabel('Sepal Width (cm)', fontsize=12)
plt.savefig('boxplot.png', dpi=300, bbox_inches='tight')
plt.show()
```

3. Pair Plot (Multi-Feature Matrix Relationships)

```python
pairplot = sns.pairplot(df, hue='species', corner=True, palette='bright')
pairplot.fig.suptitle('Pairwise Feature Matrix of Iris Dataset', y=1.02, fontsize=14, fontweight='bold')
pairplot.savefig('pairplot.png', dpi=300, bbox_inches='tight')
plt.show()
```

4. Correlation Heatmap
Purpose: Computes and displays numerical feature cross-correlations to identify linear dependencies.

```python
plt.figure(figsize=(7, 5))
# Compute correlation matrix excluding categorical 'species' column
corr_matrix = df.drop(columns=['species']).corr()

sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Feature Correlation Matrix', fontsize=14, fontweight='bold')
plt.savefig('heatmap.png', dpi=300, bbox_inches='tight')
plt.show()
```
