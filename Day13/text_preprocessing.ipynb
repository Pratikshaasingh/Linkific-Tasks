```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

1. New Text Dataset with Different Sentences

```python
corpus = [
    "Machine learning models require clean data to perform well.",
    "Deep learning architectures can understand complex patterns in language.",
    "Data science combines statistics, programming, and domain knowledge.",
    "Artificial intelligence is transforming industries across the globe."
]
```

```python
print("Original Corpus")
for i, doc in enumerate(corpus):
    print(f" {i+1}: {doc}")
```

2. Lowercasing, Tokenization, and Stopword Removal

```python
stop_words = set(stopwords.words('english'))
processed_corpus = []
```

```python
for doc in corpus:
    # Lowercasing
    lower_text = doc.lower()
    
    # Tokenization
    tokens = word_tokenize(lower_text)
    
    # Stopword Removal & filtering out punctuation
    filtered_tokens = [word for word in tokens if word.isalnum() and word not in stop_words]
    
    # Rejoin tokens back into a single string for TF-IDF
    processed_corpus.append(" ".join(filtered_tokens))
```

```python
print("\nProcessed Corpus")
for i, doc in enumerate(processed_corpus):
    print(f" {i+1}: {doc}")
```

3. TF-IDF Vectorization

```python
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(processed_corpus)
```

```python
print("\n TF-IDF Feature Names ")
print(vectorizer.get_feature_names_out())
```

```python
print("\nTF-IDF Matrix (Sparse Representation) ")
print(tfidf_matrix.toarray())
```
