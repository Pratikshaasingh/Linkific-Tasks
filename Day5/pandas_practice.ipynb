```python
import pandas as pd
import numpy as np
import seaborn as sns

# Load built-in Iris dataset directly as a Pandas DataFrame
df = sns.load_dataset('iris')

print("Iris dataset successfully loaded!")
print("Type of dataset object:", type(df))
```

```python
df.head()
```

Metadata and Missing Value Analysis

```python
# Summary of column data types and non-null values
print(" Dataset Information ")
df.info()

```

```python
# Count missing values per column
print("\n Missing Values Count ")
print(df.isnull().sum())
```

1. Filtering rows based on a single condition
# Example: Filter records where 'petal_length' is greater than 4.5 cm

```python
large_petals = df[df['petal_length'] > 4.5]

print(" Single Condition Filter ")
print(large_petals.head())

print("\n" + "="*50 + "\n")
```

2. Filtering rows using multiple conditions (AND = &, OR = |)
# Example 1 (AND): petal_length > 4.5 AND species is 'virginica'

```python
large_virginica = df[(df['petal_length'] > 4.5) & (df['species'] == 'virginica')]

print("Multiple Conditions Filter ")
print(large_virginica.head())

print("\n" + "="*50 + "\n")
```

```python
extreme_sepals = df[(df['sepal_width'] > 3.8) | (df['sepal_length'] < 4.8)]

print("Multiple Conditions Filter")
print(extreme_sepals.head())
```

Selecting a Single Column (Returns a Series)
# Example: Select only the 'petal_length' column

```python
petal_length_series = df['petal_length']

print("Single Column Selection")
print(petal_length_series.head())
```

Selecting Multiple Columns (Returns a DataFrame)
# Example: Select feature columns excluding species

```python
feature_columns = df[['sepal_length', 'sepal_width', 'petal_length']]

print("Multiple Columns Selection")
print(feature_columns.head())
```

Statistical Summary & Aggregation
Pandas makes it easy to generate statistical metrics across the whole dataset or group them by categorical variables like species.

```python
# 1. Overall Statistical Summary (describe)
print("Overall Dataset Summary Statistics")
print(df.describe())
```

```python
# 2. Individual Aggregate Functions across Numerical Columns
print("Mean values:\n", df.mean(numeric_only=True))
print("\nMedian values:\n", df.median(numeric_only=True))
print("\nStandard Deviation:\n", df.std(numeric_only=True))
```

```python
#3. Grouped Aggregation (Group by 'species')
# Calculate mean measurement for each flower species
species_mean = df.groupby('species').mean()

print("Average Measurements by Species")
print(species_mean)
```

```python
#4. Aggregating multiple statistics simultaneously
species_stats = df.groupby('species')['petal_length'].agg(['min', 'max', 'mean', 'std'])

print("Petal Length Statistics grouped by Species")
print(species_stats)
```
