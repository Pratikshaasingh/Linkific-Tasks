```python
import time
import pandas as pd
from transformers import pipeline
```

```python
print("Starting Hugging Face Multi-Model Evaluation ")
```

Task 1: Text Generation

```python
print("\n[1/3] Loading Text Generation Pipeline (Model: distilgpt2)...")
text_gen_pipeline = pipeline("text-generation", model="distilgpt2")
```

```python
gen_prompt = "Artificial Intelligence will transform software engineering by"
```

```python
start_time = time.time()
gen_result = text_gen_pipeline(
    gen_prompt, 
    max_length=60, 
    num_return_sequences=1,
    truncation=True
)
gen_latency = time.time() - start_time

gen_output_text = gen_result[0]['generated_text']
```

```python
print(f"Prompt: {gen_prompt}")
print(f"Generated Output: {gen_output_text}")
print(f"Latency: {gen_latency:.2f} seconds")
```

Task 2: Sentiment Analysis

```python
print("\n[2/3] Loading Sentiment Analysis Pipeline (Model: distilbert-base-uncased-finetuned-sst-2-english)...")
sentiment_pipeline = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")
```

```python
sentiment_input = "Exploring pre-trained Transformer models on Hugging Face makes machine learning development effortless!"
```

```python
start_time = time.time()
sent_result = sentiment_pipeline(sentiment_input)
sent_latency = time.time() - start_time
```

```python
print(f"Input Text: {sentiment_input}")
print(f"Predicted Class: {sent_result[0]['label']} (Confidence Score: {sent_result[0]['score']:.4f})")
print(f"Latency: {sent_latency:.2f} seconds")
```

Task 3: Summarization

```python
print("\n[3/3] Loading Summarization Pipeline (Model: facebook/bart-large-cnn)...")
summarizer_pipeline = pipeline("summarization", model="facebook/bart-large-cnn")
```

```python
long_text = (
    "Generative Artificial Intelligence represents a paradigm shift in software development and data science. "
    "Unlike traditional predictive machine learning models that produce discrete class labels or single numerical outputs, "
    "Generative AI systems produce entirely new artifacts including coherent text, complex software code, images, and audio. "
    "By taking advantage of the Transformer architecture and self-attention mechanisms, these models capture complex, long-range "
    "contextual relationships across vast unstructured datasets, fundamental to building intelligent modern software systems."
)
```

```python
start_time = time.time()
summary_result = summarizer_pipeline(long_text, max_length=45, min_length=20, do_sample=False)
summary_latency = time.time() - start_time
```

```python
summary_text = summary_result[0]['summary_text']
```

```python
print(f"Original Text Length: {len(long_text)} characters")
print(f"Summary Output: {summary_text}")
print(f"Latency: {summary_latency:.2f} seconds")
```
